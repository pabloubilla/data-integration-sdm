```python
from src.load_data import load_po_pa_nceas
import numpy as np

import matplotlib.pyplot as plt
```

```python
REGIONS = ["AWT", "CAN", "NSW", "SA", "SWI", "NZ"]             # regions to run
# REGIONS = ["SWI"]             # regions to run
GROUPS_BY_REGION = {
        "AWT": ["_plant", "_bird"],
        "CAN": [""],
        "NSW": ['_bat', '_bird', '_plant', '_reptile'],
        # "NSW": ["_ba", "_db", "_nb", "_ot", "_rt", "_ru", "_sr"],
        "SA" : [""],
        "SWI": [""],
        "NZ": [""]
    }  
```

```python
data_dict = {}

for region in REGIONS:
    for group in GROUPS_BY_REGION[region]:
        # 1) Load PO & PA, aligned species & covs
        X_po, Y_po, X_pa, Y_pa, species, covs = load_po_pa_nceas(
            region=region,
            group_filter=group,
            add_po_var=False,
            keep_xy=True,
            index_col=['siteid']
        )

        # save if needed later
        key = f"{region}_{group}"
        data_dict[key] = (X_po, Y_po, X_pa, Y_pa, species, covs)

        # proportions per species (shape: n_species,)
        proportion_presence_po = np.mean(Y_po, axis=0)
        proportion_presence_pa = np.mean(Y_pa, axis=0)

        # Convert species labels to strings (in case it's an Index/array/etc.)
        species_labels = [str(s) for s in species]

        # Sort species by (PA - PO) difference (optional but often helpful)
        diff = proportion_presence_pa - proportion_presence_po
        order = np.argsort(diff)  # ascending; use [::-1] for descending

        po = proportion_presence_po[order]
        pa = proportion_presence_pa[order]
        labels = [species_labels[i] for i in order]

        # Plot: side-by-side bars
        x = np.arange(len(labels))
        width = 0.45

        fig, ax = plt.subplots(figsize=(max(10, 0.35 * len(labels)), 5))
        ax.bar(x - width/2, po, width, label="PO")
        ax.bar(x + width/2, pa, width, label="PA")

        ax.set_title(f"Presence proportion by species — {region} / {group}")
        ax.set_ylabel("Proportion of presences")
        ax.set_ylim(0, 1)
        ax.set_xticks(x)
        ax.set_xticklabels(labels, rotation=90, ha="center")
        ax.legend()

        ax.grid(axis="y", linestyle="--", alpha=0.3)
        fig.tight_layout()

        plt.show()
        plt.close(fig)

        ## full means
        mean_po = np.mean(Y_po)
        mean_pa = np.mean(Y_pa)
        print(f'Mean presence PO: {mean_po:.3f}, PA: {mean_pa:.3f}')
```
